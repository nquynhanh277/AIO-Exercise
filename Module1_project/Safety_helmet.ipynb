```python
!rm yolov10n.pt  # Remove the potentially corrupted file
!wget https://huggingface.co/jameslahm/yolov10n # Download the model file again

from ultralytics import YOLOv10
MODEL_PATH = 'jameslahm/yolov10n'
model =YOLOv10.from_pretrained(MODEL_PATH)
```

```python
YAML_PATH = '../safety_helmet_dataset/data.yaml'
EPOCHS = 50
IMG_SIZE = 320
BATCH_SIZE = 128

model.train(data=YAML_PATH, epochs=EPOCHS, imgsz=IMG_SIZE, batch=BATCH_SIZE)
```

```python
from google.colab.patches import cv2_imshow

TRAINED_MODEL_PATH = '/content/yolov10/runs/detect/train/weights/best.pt'
model = YOLOv10(TRAINED_MODEL_PATH)

IMAGE_URL = 'https://www.shutterstock.com/image-photo/young-handsome-engineer-latin-man-260nw-1703786494.jpg'
CONF_THRESHOLD = 0.25

results = model.predict(IMAGE_URL, conf=CONF_THRESHOLD)
annotated_img = results[0].plot()
cv2_imshow(annotated_img)
```
