```python
### Convert color images to grayscale images using different methods

import numpy as np
import matplotlib . image as mpimg
import matplotlib.pyplot as plt
!gdown 16FkfxiUgDH0QJ7PUpwQ-Zfohv46fDcoS

img = mpimg.imread('/content/dog.jpeg')
#Convert color images to grayscale images using Lightness method
gray_img_01 = np.max(img, axis=-1, keepdims=1)/2 + np.min(img, axis=-1, keepdims=1)/2
plt.imshow(gray_img_01,cmap=plt.get_cmap('gray'))
gray_img_01[0,0]
plt.show()

#Convert color images to grayscale images using Average method
gray_img_02 = np.sum(img, axis=-1, keepdims=1)/3
plt.imshow(gray_img_02,cmap=plt.get_cmap('gray'))
gray_img_02[0,0]
plt.show()

#Convert color images to grayscale images using Luminosity method
#Setup 3 constants of Luminosity method
c1 = 0.2126
c2 = 0.7152
c3 = 0.0722
gray_img_03 = c1*img[...,0] + c2*img[...,1] + c3*img[...,2]
plt.imshow(gray_img_03,cmap=plt.get_cmap('gray'))
gray_img_03[0,0]
plt.show
```

```python
### Tabular Data Analysis

import numpy as np
import pandas as pd

!gdown 1kIjIhnCdVNnZiT2eX4pvVg4woaKEuOuh

df = pd.read_csv('/content/advertising.csv')
data = df.to_numpy()

# Find the biggest number in Sales and its index
data_sales = data[:,3]
max_data_sales = np.max(data_sales)
max_data_sales_index = np.argmax(data_sales)
max_data_sales, max_data_sales_index

# The average value of TV
data_tv = data[:,0]
avg_data_tv = data_tv.mean()
avg_data_tv

# The total number of records ín Sales >=20
sales_number_20 = np.sum(data_sales >= 20)
sales_number_20

# Calculate the average value of the Radio column with Sales column value >= 15
sales_condition = data_sales >= 15
data_radio = data[:,1]
data_radio = data_radio * sales_condition
data_radio_mean = np.sum (data_radio)/np.sum(sales_condition)
data_radio_mean

# Calculate the total values of Sales (Condition: Newpaper > the average value of all numbers in Newpaper)
data_newspaper = data[:,2]
avg_data_newspaper = np.mean(data_newspaper)
data_newspaper_condition = data_newspaper > avg_data_newspaper
data_sales_condition = data_sales * data_newspaper_condition
data_sales_total = np.sum(data_sales_condition)
data_sales_total

# Create new array that contains three values : Good, Average, Bad
avg_data_sales = data_sales.mean()
score_sales = np.where(data_sales < avg_data_sales, 'Bad', np.where(data_sales > avg_data_sales, 'Good', 'Average'))
score_sales[7:10]

# Create new array that contains three values : Good, Average, Bad
diff_sales = data_sales - avg_data_sales
diff_sales_abs = abs(diff_sales)
avg_index = np.argmin(diff_sales_abs)
avg_sale_new = data_sales[avg_index]
score_sales_new = np.where(data_sales < avg_sale_new, 'Bad', np.where(data_sales > avg_sale_new, 'Good', 'Average'))
score_sales_new[7:10]
```
