```python
#Question 8

# Import các thư viện cần thiết
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Bước 1: Tải bộ dữ liệu Iris
iris_X, iris_y = datasets.load_iris(return_X_y=True)

# Bước 2: Chia bộ dữ liệu thành tập huấn luyện và tập kiểm tra (train:test = 8:2)
X_train, X_test, y_train, y_test = train_test_split(iris_X, iris_y, test_size=0.2, random_state=42)

# Bước 3: Định nghĩa mô hình Decision Tree
dt_classifier = DecisionTreeClassifier()

# Bước 4: Huấn luyện mô hình trên tập dữ liệu huấn luyện
dt_classifier.fit(X_train, y_train)

# Bước 5: Dự đoán kết quả trên tập dữ liệu kiểm tra
y_pred = dt_classifier.predict(X_test)

# Bước 6: Tính độ chính xác của mô hình
accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy: {accuracy:.2f}')
```

```python
# Import các thư viện cần thiết
from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error

# Bước 1: Tải bộ dữ liệu Machine CPU từ OpenML
machine_cpu = fetch_openml(name='machine_cpu', version=1)
machine_data = machine_cpu.data
machine_labels = machine_cpu.target

# Bước 2: Chia bộ dữ liệu thành tập huấn luyện và tập kiểm tra (train:test = 8:2)
X_train, X_test, y_train, y_test = train_test_split(machine_data, machine_labels, test_size=0.2, random_state=42)

# Bước 3: Định nghĩa mô hình Decision Tree Regressor
tree_reg = DecisionTreeRegressor()

# Bước 4: Huấn luyện mô hình trên tập dữ liệu huấn luyện
tree_reg.fit(X_train, y_train)

# Bước 5: Dự đoán kết quả trên tập dữ liệu kiểm tra
y_pred = tree_reg.predict(X_test)

# Bước 6: Tính Mean Squared Error (MSE)
mse = mean_squared_error(y_test, y_pred)
print(f'Mean Squared Error: {mse:.2f}')
```
